```python
import pandas as pd
import os

DATASET_DIR = r'C:\Users\User\OneDrive\Documents\DATASETdatavis'
OUTPUT_FILE = os.path.join(DATASET_DIR, 'Gasto_OPDS_2018_2026.csv')

years = range(2018, 2027)
dfs = []

for year in years:
    path = os.path.join(DATASET_DIR, f'{year}-Gasto-OPDS.csv')
    df = pd.read_csv(path, encoding='utf-8', low_memory=False)
    df['anio_archivo'] = year
    dfs.append(df)
    print(f'{year}: {len(df):,} filas, {df.shape[1]} columnas')

combined = pd.concat(dfs, ignore_index=True)
print(f'\nTotal combinado: {len(combined):,} filas, {combined.shape[1]} columnas')
```

```python
combined.to_csv(OUTPUT_FILE, index=False, encoding='utf-8-sig')
print(f'Archivo guardado en: {OUTPUT_FILE}')
```
